```python
import cv2
import numpy as np
from tensorflow.keras.models import load_model
```

```python
# model = load_model("fruit_classification.keras")
model = load_model("fruit_classification_1.keras")
```

```python
IMG_SIZE = (128,128)

Class_label = [
    "Apple",
    "Banana",
    "Grape",
    "Mango",
    "Strawberry"
]
cap = cv2.VideoCapture(0)  ### 0 for camera

if not cap.isOpened():
    print("Error : Could not open camera")
else:
    print("Camera Opened Successfully")
```

```python
def preprocess_and_predict(frame, model, img_size, class_labels):

    resized_frame = cv2.resize(frame, img_size)

    resized_frame = cv2.cvtColor(resized_frame, cv2.COLOR_BGR2RGB)

    normalized_frame = (resized_frame.astype('float32') / 255.0)

    input_tensor = np.expand_dims(normalized_frame, axis=0)

    prediction = model.predict(input_tensor, verbose=0)

    predicted_index = np.argmax(prediction[0])

    predicted_class = class_labels[predicted_index]

    return predicted_class
```

```python
while True:
    ret, frame = cap.read()
    if not ret:
        break
    predicted_label = preprocess_and_predict(frame, model, IMG_SIZE,Class_label)
    text = f"{predicted_label}"
    cv2.putText(frame, text,(50,50), cv2.FONT_HERSHEY_SIMPLEX, 1,(0,0,255), 2)
    cv2.imshow('Live CNN Classification', frame)
    if cv2.waitKey(1) & 0xff == ord('q'):
        cap.release()
        cv2.destroyAllWindows()
        break
```
